```python
from pathlib import Path 
import pandas as pd
import numpy as np 

path = Path.cwd().parent/'data'/'raw'

df_2526 = pd.read_csv(path/'E0_2526.csv')

df_2526

```

Wybiore teraz kolumny które mi posłużą do analizy i wytrenowania modelu.


```python
SELECTED_COLUMNS = [
    # Podstawowe informacje i wynik
    'Date', 'HomeTeam', 'AwayTeam', 
    'FTHG', 'FTAG', 'FTR', 
    'HTHG', 'HTAG', 'HTR',
    
    # Statystyki meczowe
    'HS', 'AS',     # Strzały
    'HST', 'AST',   # Strzały celne
    'HF', 'AF',     # Faule
    'HC', 'AC',     # Rzuty rożne
    'HY', 'AY',     # Żółte kartki
    'HR', 'AR',     # Czerwone kartki
    
    # Kursy bukmacherskie (rynek 1X2)
    'AvgH', 'AvgD', 'AvgA',    # Średnia rynkowa
    'B365H', 'B365D', 'B365A', # Bet365
]
```

```python
new_df_2526 = df_2526[SELECTED_COLUMNS]

new_df_2526.head(10)
```

```python
new_df_2425 = df_2425[SELECTED_COLUMNS]
new_df_2425.head(10)
```

```python
new_df_2324 = df_2324[SELECTED_COLUMNS]
new_df_2324.head(10)
```

```python
print(new_df_2324.isna())
print(new_df_2425.isna())
print(new_df_2526.isna())
```

```python
print(new_df_2324.isna().sum().sum())
print(new_df_2526.isna().sum().sum())
print(new_df_2425.isna().sum().sum())

```
